# Events and holidays

有两种方法可以向 NeuralProphet 添加特殊事件：自定义事件和预定义国家假日。两者都可用于模拟其他组件未捕获的特殊场合和事件。这些特殊事件有助于提高模型的性能和可解释性。

首先我们研究国家假日，稍后我们将探索自定义事件。

```python
import pandas as pd
from neuralprophet import NeuralProphet, set_log_level

# Load the dataset for tutorial 4 with the extra temperature column
df = pd.read_csv('./data/SF_hospital_load.csv')
df.head()

# Disable logging messages unless there is an error
set_log_level("ERROR")

# Default model
m = NeuralProphet()
m.set_plotting_backend("plotly-static")

# Continue training the model and making a prediction
metrics = m.fit(df)
forecast = m.predict(df)
m.plot(forecast)
```

NeuralProphet 支持广泛的国家和地区。完整列表可在 [pyholidays 文档](https://python-holidays.readthedocs.io/en/latest/) 中找到，该文档内部用于国家/地区假日支持。

可以使用 `add_country_holidays` 函数轻松添加国家/地区假日：

```python
m = NeuralProphet()
m.set_plotting_backend("plotly-static")

m = m.add_country_holidays("US")

metrics = m.fit(df)
forecast = m.predict(df)
m.plot(forecast)
```

特殊事件或节假日的影响可能偶尔发生。此类事件的建模类似于未来回归量，每个事件 $e$ 都是一个二元变量 $e ∈ [0, 1]$，表示该事件是否在特定日期发生。

## Custom events

自定义事件与国家假期非常相似，但可用于模拟任何特殊事件。唯一的区别是，事件名称和日期需要由用户指定。此外，用户必须将事件包含在训练数据中或相应地准备数据集。我们将在以下示例中探讨后一种选项，以说明自定义事件数据集应是什么样子。

```python
# https://eswd.eu/cgi-bin/eswd.cgi

df_events = pd.DataFrame(
    {
        "event": "extreme_weather",
        "ds": pd.to_datetime(
            [
                "2018-11-23",
                "2018-11-17",
                "2018-10-28",
                "2018-10-18",
                "2018-10-14",
            ]
        ),
    }
)

m = NeuralProphet()
m.set_plotting_backend("plotly-static")

m.add_events("extreme_weather")

df_all = m.create_df_with_events(df, df_events)

metrics = m.fit(df_all)
forecast = m.predict(df_all)
m.plot(forecast)
```
