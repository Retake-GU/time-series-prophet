#### NeuralProphet 模型的核心概念是其模块化可组合性。 该模型由模块组成，每个模块都为预测贡献了一个附加组件。 大多数组件也可以配置为按趋势缩放以获得乘法效果。 每个模块都有自己的个人输入和建模过程。 但是，所有模块都必须产生 h 个输出，其中h 定义了一次预测未来的步数。

$$\text{y}(t) = T(t) + S(t) + E(t) + F(t) + A(t) + L(t)$$

T(t) = 时间 t 的趋势

S(t) = 时间 t 的季节性影响

E(t) = 时间 t 的事件和假日效应

F(t) = 未来已知外生变量在时间 t 的回归效应 future regressors

A(t) = 基于过去观察的时间 t 的自回归效应 auto regression

L(t) = t 时刻外生变量滞后观测的回归效应 lagged regressors

所有模型组件模块都可以单独配置和组合以组成模型。 如果所有模块都关闭，则仅安装一个静态偏移参数作为趋势分量。 默认情况下，仅激活趋势和季节性模块。

使用了$MAPE$公式：
$$\text{MAPE} = \frac{1}{n} \sum_{i=1}^{n} \left| \frac{y_i - \hat{y}_i}{y_i} \right| \times 100
$$
其中${y_i}$是实际值,$\hat{y}_i$是预测值，$n$是数据点的总数。
